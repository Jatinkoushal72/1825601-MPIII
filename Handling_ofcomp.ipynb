```python
##Handling Of Complex numbers
##1==Nth root of complex Number
import cmath
import math

def nth_root(z, n):
  r = abs(z)
  theta = cmath.phase(z)
  roots = []

  for k in range(n):
    root_r = r**(1/n)
    root_theta = (theta + 2*cmath.pi*k)/n

    root = cmath.rect(root_r, root_theta)
    roots.append(f"Root {k+1}: {root}")
  return "\n".join(roots)

z = complex(5, 4)
N_th_root = nth_root(z, 4)
print(N_th_root)
```

```python
##Complex Transformation
##2==
import matplotlib.pyplot as plt
import numpy as np

#Defining a triangle using complex numbers.
z = np.array([1 + 1j, 3 + 1j, 2 + 3j, 1 + 1j])


translated = z + (2 + 1j)
scaled = z * 1.5
rotated = z * np.exp(1j * np.pi / 4)
reflected = z.conjugate()
plt.scatter(z.real, z.imag)
plt.scatter(translated.real, translated.imag)
plt.scatter(scaled.real, scaled.imag)
plt.scatter(rotated.real, rotated.imag)
plt.scatter(reflected.real, reflected.imag)

plt.plot(z.real, z.imag,label='Original')
plt.plot(translated.real, translated.imag, label='Translated')
plt.plot(scaled.real, scaled.imag, label='Scaled')
plt.plot(rotated.real, rotated.imag, label='Rotated (45°)')
plt.plot(reflected.real, reflected.imag, label='Reflected')


plt.axhline(0, color='black')
plt.grid(True)
plt.legend()
plt.show()


```

```python
## Visualisation of mapping from z-plane to w-plane
##3==
import matplotlib.pyplot as plt
import numpy as np

#Defining a triangle in z-plane
z = np.array([1 + 0j, 2 + 0j, 1.5 + 1j, 1 + 0j])


w = z**2

plt.plot(z.real, z.imag, color='red')
plt.plot(w.real, w.imag, color = 'blue')
plt.title('w-plane (Output: w = z²)')

plt.grid(True)

plt.show()
```
