```python
##Finite Difference Method
import numpy as np
import matplotlib.pyplot as plt
N = 100
L = 2*np.pi
h = L/N
y_st = 0
y_end = 0.1
t = np.linspace(1,L,N+1)
X = np.zeros((N-1,N-1))
D = np.zeros(N-1)
for i in range(N-1):
  X[i,i] = h**2 - 2
  if i>=0:
    X[i,i-1] = 1
  if i<N-2:
    X[i,i+1] = 1


D[0]= y_st
D[-1]= -y_end

S = np.linalg.solve(X,D)
print(X)
print(S)
plt.plot(t[1:-1],S)
plt.show()
```
