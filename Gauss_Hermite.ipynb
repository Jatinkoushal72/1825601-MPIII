```python
##Gauss Hermite
import numpy as np
def func(x):
    return (np.exp(x**2))*(1/(x**2 + 1))

x = np.array([-1.650680,-0.524648,1.650680,0.524648])
w = np.array([0.081313,0.804914,0.081313,0.804914])

IN =np.sum(w*func(x))

print("Gauss Hermite integration:", IN)
```
