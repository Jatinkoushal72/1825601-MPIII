```python
##Gauss Legendre
import numpy as np
def func(x):
  return (1/(x**2 + 1))
w = np.array([0.347855,0.652145,0.652145,0.347855])
x = np.array([-0.861136,-0.339981,0.339881,0.861136])

I = np.sum(w*func(x))

print(I)

```
