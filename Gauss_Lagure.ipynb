```python
##Gauss Lagure
import numpy as np
def func(x):
    return (np.exp(x))*(1/(x**2 + 1))

x = np.array([0.322548,1.74576,4.53662,9.39507])
w = np.array([0.603154,0.357419,0.038888,0.000539])

IN = np.sum(w*func(x))

print("Gauss Lagure integration:", IN)
```
