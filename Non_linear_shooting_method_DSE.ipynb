```python
###Shooting method with non-linear equation
#Co-efficients of variables in the given equation.
p = 0
q = 0
r = -6


def func(x, y1, y2):
    return p - q*y2 - r*(y1**2)


def rk4(xi, y1_i, y2_i, xf, n):
    h = (xf - xi)/n
    xarr, y1_arr, y2_arr = [], [], []
    xarr.append(xi)
    y1_arr.append(y1_i)
    y2_arr.append(y2_i)
    for i in range(1, n+1):
        k1_y1 = y2_i
        k1_y2 = func(xi, y1_i, y2_i)

        k2_y1 = y2_i + (h/2)*k1_y2
        k2_y2 = func(xi + h/2, y1_i + (h/2)*k1_y1, y2_i + (h/2)*k1_y2)

        k3_y1 = y2_i + (h/2)*k2_y2
        k3_y2 = func(xi + h/2, y1_i + (h/2)*k2_y1, y2_i + (h/2)*k2_y2)

        k4_y1 = y2_i + h*k3_y2
        k4_y2 = func(xi + h, y1_i + (h)*k3_y1, y2_i + (h)*k3_y2)

        xi = xi + h
        y1_i = y1_i + (k1_y1 + 2*k2_y1 + 2*k3_y1 + k4_y1)*(h/6)
        y2_i = y2_i + (k1_y2 + 2*k2_y2 + 2*k3_y2 + k4_y2)*(h/6)
        xarr.append(xi)
        y1_arr.append(y1_i)
        y2_arr.append(y2_i)
    return xarr, y1_arr, y2_arr


x0, xf = 1,2
y1_i = 1
y_f_target = 0.25


def shoot_error(guess_y2):
    xres, y1res, y2res = rk4(x0, y1_i, guess_y2, xf, 100)
    return (y1res[-1]) - y_f_target


G1 = 1
G2 = 2
error1 = shoot_error(G1)

tol = 1e-6
while abs(error1) > tol:
    error2 = shoot_error(G2)
    next_G = G2 - error2*(G2-G1)/(error2 - error1)

    G1, G2 = G2, next_G
    error1 = error2

correct_y2_i = G1
xres, y1res, y2res = rk4(x0, y1_i, correct_y2_i, xf, 100)
print("Correct first derivative y'(x0):", correct_y2_i)
print("Final y(xf):", y1res[-1])
```
