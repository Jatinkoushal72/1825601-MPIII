```python
#Rotation of a matrix
import numpy as np
import matplotlib.pyplot as plt
import math as mt
r = np.array([[np.cos(mt.radians(5)),np.sin(mt.radians(5))],[-np.sin(mt.radians(5)),np.cos(mt.radians(5))]])
a = np.array([6,6])
plt.scatter(a[0],a[1])
for i in range(72):
  f=r@a
  plt.scatter(f[0],f[1])
  a = f
plt.show()
```
