```python
##FFT And IFT                                  Dated:12/10/26
import cmath
import numpy as np

x = np.array([1,2,3,4])
N = len(x)
F_k = np.zeros(N,dtype=complex)
def dft(x):
    return [sum(x[n]*cmath.exp(1j*2*cmath.pi*k*n/N) for n in range(N))
            for k in range(N)]


result1 = dft(x)
print(result1)

def Ift(x):
    N = len(x)
    return [sum(x[n]*cmath.exp(1j*2*cmath.pi*k*n/N) for n in range(N))
            for k in range(N)]
result2 = Ift(x)
print(result2)
```
